```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Extract the ZIP file
import zipfile

with zipfile.ZipFile("AB_NYC_2019.csv.zip", "r") as zip_ref:
    zip_ref.extractall(".")

print("ZIP file extracted successfully!")

# Load the CSV file
df = pd.read_csv("AB_NYC_2019.csv")

print("Dataset loaded successfully!")
print("Shape of dataset:", df.shape)
```

```python
df.head()
```

```python
# Make a copy
df_clean = df.copy()

# Fill missing reviews per month
df_clean['reviews_per_month'] = df_clean['reviews_per_month'].fillna(0)

# Remove rows with missing listing names
df_clean = df_clean.dropna(subset=['name'])

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Convert last_review to datetime
df_clean['last_review'] = pd.to_datetime(
    df_clean['last_review'],
    errors='coerce'
)

# Remove zero-price listings
df_clean = df_clean[df_clean['price'] > 0]

print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
```

```python
Q1 = df_clean['price'].quantile(0.25)
Q3 = df_clean['price'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
df_clean = df_clean[
    (df_clean['price'] >= lower_bound) &
    (df_clean['price'] <= upper_bound)
]
```

```python
df_clean['price'].describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

df_clean.groupby('neighbourhood_group')['price'].mean().sort_values().plot(
    kind='bar'
)

plt.title('Average Airbnb Price by Neighbourhood Group')
plt.xlabel('Neighbourhood Group')
plt.ylabel('Average Price ($)')

plt.show()
```
